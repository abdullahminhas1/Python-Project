**Load and Preview the Data**


```python
import pandas as pd
import numpy as np

df = pd.read_excel("scraped_zameen_data.xlsx")
df_info = df.info()
df_shape = df.shape

df.head()
```

**Remove Empty columns and Duplicate Rows**

```python
df_cleaned = df.dropna(axis=1, how='all')
print(df_cleaned.columns.tolist())

duplicate_rows = df_cleaned.duplicated().sum()
print("Duplicate Rows:", duplicate_rows)

df_cleaned = df_cleaned.drop_duplicates()
print("Cleaned Data:", df_cleaned)

display(df_cleaned.head())
```

**Clean Messy Columns (Price & Area Formatting)**

***Price***

```python
import numpy as np
import re

def clean_price(value):
    if isinstance(value, (int, float)):
        return float(value)

    if isinstance(value, str):
        text = value.lower().replace("pkr", "").replace(",", "").strip()
        num_match = re.findall(r"\d+\.?\d*", text)
        if not num_match:
            return np.nan
        num = float(num_match[0])
        if "crore" in text:
            return num * 10_000_000
        elif "lakh" in text:
            return num * 100_000
        elif "thousand" in text:
            return num * 1_000
        else:
            return num
    return np.nan

df_cleaned['Price_Cleaned'] = df_cleaned['Price'].apply(clean_price)
df_cleaned[['Price', 'Price_Cleaned']].head(20)
```

***Area***

```python
def clean_area(value):
    if pd.isnull(value):
        return np.nan
    if isinstance(value, (int, float)):
        return float(value)

    text = str(value).lower().strip()
    nums = re.findall(r'\d+\.?\d*', text)
    if not nums:
        return np.nan
    num = float(nums[0])

    if 'marla' in text:
        return num * 225
    elif 'kanal' in text:
        return num * 4500
    elif 'sq. yd' in text or 'sq yd' in text:
        return num * 9
    elif 'sqft' in text or 'sq. ft' in text or 'sq ft' in text:
        return num
    else:
        return np.nan

df_cleaned['Area_Sqft'] = df_cleaned['Area'].apply(clean_area)
df_cleaned[['Area', 'Area_Sqft']].head()
```

**Missing Values Treatment**

```python
#Missing values count across cols in DataSet
missing_values = df_cleaned.isnull().sum()
print("Total Missing Values per Column:")
display(missing_values)

#Sorting missing values
missing_valuesPerCol = missing_values.sort_values(ascending=False)
print("Missing Values by Column (sorted):")
display(missing_valuesPerCol)

#Sorting cols in DataSet accordingly
num_cols = df_cleaned.select_dtypes(include=['int64','float64']).columns
cat_cols = df_cleaned.select_dtypes(include=['object']).columns
print(f"Numerical Columns: {list(num_cols)}")
print(f"Categorical Columns: {list(cat_cols)}")

#Filling numerical cols with mean
df_cleaned[num_cols] = df_cleaned[num_cols].fillna(df_cleaned[num_cols].mean())

#Filling categorical cols with mode
for col in cat_cols:
    mode_value = df_cleaned[col].mode()
    if not mode_value.empty:
        df_cleaned[col] = df_cleaned[col].fillna(mode_value[0])


#Final Summary
print("Remaining missing values per column after imputation:")
display(df_cleaned.isnull().sum())

total_missing_after = df_cleaned.isnull().sum().sum()
print(f"Total remaining missing values in the dataset: {total_missing_after}")
```

**Data Cleaning & Consistency**

```python
!pip install fuzzywuzzy[speedup] --quiet
from fuzzywuzzy import process

cities_data = df_cleaned['City'].unique()
print(cities_data)

top_cities = df_cleaned['City'].value_counts().head(20).index.tolist()

def standardize_city(city):
    if pd.isnull(city):
        return None

    city = city.strip()
    match = process.extractOne(city, top_cities)

    if match is None:
        return city.title()

    matched_name, score = match
    if score >= 85:
        return matched_name
    else:
        return city.title()

df_cleaned['City_Standard'] = df_cleaned['City'].apply(standardize_city)

display(df_cleaned[['City', 'City_Standard']].drop_duplicates().head(15))

```

**Detect and Correct Inconsistent Property Types**

```python
df_cleaned['Type_Cleaned'] = df_cleaned['Type'].str.strip().str.lower()

type_mapping = {
    'flat': 'Apartment',
    'apartment': 'Apartment',
    'house': 'House',
    'upper portion': 'Portion',
    'lower portion': 'Portion',
    'portion': 'Portion',
    'penthouse': 'Penthouse',
    'room': 'Room',
    'farm house': 'Farmhouse',
    'villa': 'Villa',
    'office': 'Commercial',
    'shop': 'Commercial',
    'plaza': 'Commercial',
    'warehouse': 'Commercial',
    'factory': 'Commercial',
}

df_cleaned['Type_Standard'] = df_cleaned['Type_Cleaned'].map(type_mapping).fillna('Other')

df_cleaned['Type_Standard'].value_counts()
df_cleaned[['Type', 'Type_Standard']].drop_duplicates().head(20)
```

**Outlier Detection and Removal**

```python
def outlier_removal(df):
    numeric_cols = df.select_dtypes(include='number').columns
    for column in numeric_cols:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df = df[(df[column] >= lower) & (df[column] <= upper)]
    return df

df_cleaned = outlier_removal(df_cleaned)
print(f"Dataset shape after outlier removal: {df_cleaned.shape}")
```

## **Conclusion**

The exploratory data analysis of Zameen.com property listings depicted a vivid picture of Pakistan’s online real estate landscape. By cleaning inconsistent price formats, converting mixed area units into standard square footage, and exploring pricing patterns across cities and property types, major factors were identified which influence property values across.

The analysis shows that location, area, property type, and bedroom count are the strongest factors affecting price. Big cities dominate the luxury and high-value segment, while mid-tier cities provide more affordable and stable investment opportunities. The dataset also highlights the diversity of the Pakistani property market across the board.
